Sadly litellm isn't compatible with groq, so I needed to build a workaround in order to use it. It seems to work...?

```python
import dspy
import os
from dotenv import load_dotenv
```

```python
load_dotenv()
groq_api_key = os.getenv("GROQ_API_KEY")
gpt_oss_120b = dspy.LM("groq/openai/gpt-oss-120b", api_key=groq_api_key)
dspy.configure(lm=gpt_oss_120b, adapter=dspy.JSONAdapter)
```

```python
questions = ["1+1?", "2+2?", "3+3?"]
examples = [
    dspy.Example(
        question=q
    ).with_inputs("question")
    for q in questions
]
examples
```

```python
mathbot = dspy.Predict("question -> answer")
```

```python
answer = await mathbot.abatch(examples=examples)
answer
```
